# Interactive Travel UI Flow (Step-by-Step)

## Step 1: Destination Selection
**UI:** Three-way split-screen with images and highlights.  
**Action:** ➡️ Student clicks a destination.  
**Sample Destinations:**
- San Diego: Balboa Park, Coronado Bridge, Palm Trees
- Los Angeles: Hollywood Sign, Walk of Fame, Griffith Observatory
- San Francisco: Golden Gate Bridge, Pier 39, Alcatraz Island
- Portland: Crater Lake, Multnomah Falls, Japanese Garden
- Seattle: Space Needle, Pike Place Market, Lumen Field

## Step 2: Travel Animation
**UI:** Car drives to selected city with map-style path lines.  
**Goal:** Immersive travel transition.

## Step 3: Scenic View Animation
**UI:** Close-up scenic shot (e.g., Golden Gate skyline, Balboa Park gardens, Multnomah Falls).  
**Style:** Cinematic pan, ambient sound, subtle motion.

## Step 4: Learn Coding
**UI:** 🎓 Button: "Click here to learn how we coded this."  
**Functionality:** Opens side panel/overlay with annotated code snippets or mini tutorials.

## Step 5: Coding Quiz
**UI:** Short multiple-choice or coding-input quiz.  
**AI:** Evaluates answers using rubric (logic, syntax, creativity), gives instant feedback.  
**Goal:** Reinforce understanding through practice.

## Step 6: Social Sharing
**UI:** 📸 Camera icon: "Click here to post on Codegram."  
**Functionality:** Share screenshots or progress badges for peer engagement.
